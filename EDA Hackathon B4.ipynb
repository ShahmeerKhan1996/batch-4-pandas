```python
import pandas as pd
import warnings
import seaborn as sns
import matplotlib.pyplot as plt
warnings.filterwarnings('ignore')
```

Load the dataset using Python Library Pandas

```python
df = pd.read_csv("ecommerce_dataset.csv")
df
```

Check types of data

```python
df.dtypes
```

Check duplicate rows



```python
df.duplicated()
```

```python
df[df.duplicated()]
```

Drop duplicate records



```python
df = df.drop_duplicates()
df
```

Check missing and null vales



```python
df.isnull()
```

```python
df.isnull().sum()
```

Replace missing values with median

```python
df['City Code'] = df['City Code'].fillna(df['City Code'].median())
df['Order Priority Code'] = df['Order Priority Code'].fillna(df['Order Priority Code'].median())
```

```python
df.isnull().sum()
```

Drop Irrelevant Columns


```python
df = df.drop(['Latitude','Longtitude'], axis=1)
df
```

Rename Columns

```python
df = df.rename(columns={"City Code": "City", "Order Priority Code": "Order Priority"})
df
```

Top 5 Best Selling Products (Bar Chart) Using Seaborn

```python
sns.set_theme(style="whitegrid")

top_products = (
    df.groupby("product")["total"]
    .sum()
    .sort_values(ascending=False)
    .head(5))

top_df = top_products.reset_index()

plt.figure(figsize=(7,4))
sns.barplot(data=top_df, x="product", y="total")

plt.title("Top 5 Best Selling Products")
plt.xlabel("Product")
plt.ylabel("Total Sales")
plt.tight_layout()
plt.show()
```

Observation: Best selling product is headphones, need to maintain the growth

Payment Method Distribution (Pie Chart) Using Matplotlib

```python
payment_counts = df["payment_method"].value_counts()

plt.figure(figsize=(6,6))
payment_counts.plot(kind="pie", autopct="%1.1f%%")
plt.title("Payment Method Distribution")
plt.ylabel("")
plt.tight_layout()
plt.show()
```

Observation: Customers are preffering to pay digitally instead of cash

Month-wise Sales (Line Chart) Using Matplotlib

```python
month_sales = df.groupby("Month")["total"].sum()

months_order = [
    "January","February","March","April","May","June","July","August",
    "September","October","November","December"
]
month_sales = month_sales.reindex(months_order, fill_value=0)

plt.figure(figsize=(8,4))
plt.plot(month_sales.index, month_sales.values, marker='o', linestyle='-', color='blue')
plt.title("Month-wise Total Sales")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

Observations: Sales dropping by signifcant margin after May then again start peaking in December, need to analyze root cause between July to Nov

Show monthly Sales by Category side by side bar chart using matplotlib

```python
category_sales = df.groupby("category")["total"].sum().reset_index()
month_category_sales = df.groupby(["Month", "category"])["total"].sum().unstack(fill_value=0)

month_order = [
    "January","February","March","April","May","June","July","August",
    "September","October","November","December"
]

month_category_sales = month_category_sales.reindex(month_order, fill_value=0)

month_category_sales.plot(kind='bar', figsize=(10,5))
plt.title("Monthly Sales by Category")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.legend(title="Category")
plt.tight_layout()
plt.show()
```

Observation:

- Accessories sales were consistent between Jan to Apr then decling till Oct
- Fashion items beat electronics and accesories sales in Dec
- Electronics sales was high over the year in Oct
